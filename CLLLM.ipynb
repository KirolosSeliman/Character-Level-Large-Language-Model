# Character Level Large Language Model


```python
words = open('names.txt').read().splitlines()
```

```python
words[:5]
words.__len__()
```

```python
for w in words:
    chs = ['<S>'] + list(w) + ['<E>'] # wrapping the words with a start and end markers
    for ch1, ch2 in zip(w, w[1:]):
        print(ch1, ch2)
```
